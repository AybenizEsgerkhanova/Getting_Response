```python
OPENROUTER_API_KEY = os.getenv("OPENRouter_API_KEY")
```

```python
openrouter_client = OpenAI(
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)
```

```python
response = openrouter_client.responses.create(model="openrouter/auto",
                       input = "Solve math problems")
```

```python
system_instruction = """
You are an expert Python programming instructor.
Strict rules:
1. Answer ONLY in English.
2. DO NOT answer any questions or topics unrelated to Python programming. If asked about other topics, politely decline.
"""

prompt = "Explain Python Decorators with a simple code example."
```

```python
response = openrouter_client.chat.completions.create(
    model="nvidia/nemotron-3-ultra-550b-a55b:free",
    messages=[
        {"role": "system", "content": system_instruction},
        {"role": "user", "content": prompt}
    ],
    max_tokens=300
)

print(response.choices[0].message.content)
```
